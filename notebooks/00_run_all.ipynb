# 00 · Run the Whole Pipeline

Executes notebooks 01–10 in order. **Each notebook runs in its own fresh kernel**, which is shut down when it
finishes, so memory is released between steps (important on an 8 GB machine). Outputs are saved back into each
notebook, so you can open any of them afterwards to read the results.

**Usage:** set `FROM` / `TO` below (e.g. `"04"`, `"10"` to re-run only part of the pipeline), then run this cell.
Progress is printed after each notebook. If a notebook fails, the run stops, the failing notebook is saved with its
outputs up to the error, and the error is shown here.

Approximate runtime on the full data: 02 ≈ 2 min, 07 ≈ 20–30 min, the rest a few minutes each (≈ 1 hour in total).

```python
FROM, TO = "01", "10"          # first and last notebook to run (two-digit prefixes)
KERNEL = "homecredit"          # Jupyter kernel name of the project environment (see README)

import re
import time
from pathlib import Path

try:
    import nbformat
    from nbclient import NotebookClient
    from nbclient.exceptions import CellExecutionError
except ImportError as e:
    raise ImportError("Install the runner dependencies in the project environment: pip install nbclient nbformat") from e

NB_DIR = Path.cwd().resolve()
NB_DIR = NB_DIR / "notebooks" if (NB_DIR / "notebooks").is_dir() else NB_DIR
todo = sorted(p for p in NB_DIR.glob("[0-9][0-9]_*.ipynb") if p.name[:2] != "00" and FROM <= p.name[:2] <= TO)
print("will run:", [p.name for p in todo])

t_all = time.time()
for path in todo:
    t0 = time.time()
    print(f"\n▶ {path.name} ...", flush=True)
    nb = nbformat.read(path, as_version=4)
    client = NotebookClient(nb, timeout=None, kernel_name=KERNEL, resources={"metadata": {"path": str(NB_DIR)}})
    try:
        client.execute()
    except CellExecutionError as err:
        nbformat.write(nb, path)                     # keep outputs up to the failing cell
        print(f"✗ {path.name} failed after {(time.time() - t0) / 60:.1f} min — open it to see the full traceback.")
        last_lines = re.sub(r"\x1b\[[0-9;]*m", "", str(err)).strip().splitlines()[-6:]   # strip colour codes
        raise RuntimeError(f"{path.name} failed:\n" + "\n".join(last_lines)) from None
    nbformat.write(nb, path)
    print(f"✓ {path.name} done in {(time.time() - t0) / 60:.1f} min", flush=True)
print(f"\nAll done in {(time.time() - t_all) / 60:.1f} min.")
```
